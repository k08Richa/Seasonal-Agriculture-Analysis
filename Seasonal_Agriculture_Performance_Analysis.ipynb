## Data Cleaning

Three columns contained missing values:
- Rainfall_mm: 48 missing (1.2%) — filled with median
- Soil_Moisture_pct: 40 missing (1.0%) — filled with median
- Yield_Tonnes_Ha: 32 missing (0.8%) — filled with median

Median was chosen over mean because agricultural data often contains
outliers that skew the average. Filling with median preserves the
central tendency without distortion.

```python
# Fill missing values with median
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())
```

```python
# Average yield, revenue, cost, profit by crop
crop_summary = df.groupby('Crop')[['Yield_Tonnes_Ha', 'Revenue_INR', 'Total_Cost_INR', 'Profit_INR']].mean().round(2).sort_values('Profit_INR', ascending=False)
print('Average Financial Performance by Crop:')
print(crop_summary)
```

```python
# Average yield and profit by season
season_summary = df.groupby('Season')[['Yield_Tonnes_Ha', 'Profit_INR', 'Rainfall_mm', 'Water_Efficiency_t_per_1000m3']].mean().round(2)
print('Average Performance by Season:')
print(season_summary)
```

```python
# Average yield and water efficiency by irrigation method
irrigation_summary = df.groupby('Irrigation_Method')[['Yield_Tonnes_Ha', 'Water_Efficiency_t_per_1000m3', 'Profit_INR']].mean().round(2).sort_values('Yield_Tonnes_Ha', ascending=False)
print('Average Performance by Irrigation Method:')
print(irrigation_summary)
```

## Key Findings

1. **Crop Performance**: Sugarcane has the highest average yield by a
   significant margin compared to all other crops which
   average below 3 Tonnes/Ha. For profitability, Sugarcane and Chilli
   show positive average profit while Maize, Rice, and Wheat show
   negative average profit - meaning costs exceed revenue for these
   commonly grown crops.

2. **Season Impact**: Kharif season produces the highest average yield
  and is the most profitable season. Rabi season shows moderate positive profit. Zaid
   (summer) season has the lowest yield and records average losses.

3. **Irrigation Efficiency**: Rainfed farms show the highest water
   efficiency ratio, followed by Drip irrigation. Flood irrigation is
   the least water-efficient method - using the most water for the
   lowest output per unit. Where managed irrigation is needed, Drip
   is the most efficient choice.

4. **Rainfall & Yield**: The scatter plot shows no strong linear
   relationship between rainfall and yield. Farms across the full
   rainfall range (100mm to 1400mm) show similar yield distributions,
   suggesting rainfall alone does not determine farm output.

5. **Seed Quality**: The scatter plot shows no clear directional pattern
   between seed quality score and yield. High and low yields appear
   across all seed quality levels, indicating yield is determined by
   multiple factors together rather than seed quality alone.

6. **Disease & Pest Risk**: Risk is normally distributed across farms
   with the majority falling between 35-60%. Very few farms fall below
   15% or above 75%, suggesting disease and pest risk is a universal
   concern across all crops and regions rather than being specific to
   any particular area.

## Conclusion

This analysis of 4,000 farm records from across India reveals that crop yield
and profitability are influenced by a combination of environmental, operational,
and input quality factors. Irrigation method choice - particularly rainfed irrigation -
significantly impacts water efficiency. Seed quality shows a consistent positive
relationship with yield outcomes. Seasonal variation affects both rainfall availability
and crop performance, with some seasons consistently outperforming others.

These findings can support agricultural planning decisions for farmers, policymakers,
and agri-businesses looking to optimise resource allocation, improve yield, and
reduce operational losses.
